### Step 1: Import Required Libraries
We import `pandas` for data manipulation, `numpy` for mathematical operations, and `matplotlib.pyplot` for graphical descriptive analytics.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set plot style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
print('Libraries imported successfully!')
```

### Step 2: Load the Real-World Dataset
We load `post_finance_metrics.csv` containing 10,000 records of investor tweets and financial metrics across 37 equities in 2020.

```python
# Load dataset
data = pd.read_csv('../analysis/post_finance_metrics.csv')
print(f'Dataset Shape: {data.shape[0]:,} Rows × {data.shape[1]} Columns')
data[['date', 'ticker', 'emo_label', 'senti_label', 'word_count', 'char_count', 'same_day_return', 'intraday_range']].head()
```

### Step 3: Textual Descriptive Analytics Using Built-in Functions

#### 3.1 Measures of Central Tendency
- **Mean (`data.mean()`)**: Arithmetic average of the distribution.
- **Median (`data.median()`)**: 50th percentile (middle value), robust to outliers.
- **Mode (`data.mode()`)**: Most frequently occurring value in the dataset.

```python
num_cols = ['word_count', 'char_count', 'same_day_return', 'intraday_range', 'fwd_vol_5d']

print('--- Measures of Central Tendency ---')
central_tendency = pd.DataFrame({
    'Mean': data[num_cols].mean().round(4),
    'Median': data[num_cols].median().round(4),
    'Mode': data[num_cols].mode().iloc[0].round(4)
})
central_tendency
```

#### 3.2 Measures of Dispersion
- **Minimum (`data.min()`)** & **Maximum (`data.max()`)**: The smallest and largest observations.
- **Range (`max - min`)**: Total span of data.
- **Variance (`data.var()`)**: Average squared deviation from the mean.
- **Standard Deviation (`data.std()`)**: Average spread in original units.

```python
print('--- Measures of Dispersion ---')
dispersion = pd.DataFrame({
    'Minimum': data[num_cols].min().round(4),
    'Maximum': data[num_cols].max().round(4),
    'Range': (data[num_cols].max() - data[num_cols].min()).round(4),
    'Variance': data[num_cols].var().round(6),
    'Standard Deviation': data[num_cols].std().round(4)
})
dispersion
```

#### 3.3 Shape of Distribution
- **Skewness (`data.skew()`)**: Asymmetry of distribution (0 = symmetric, >0 = right/positive skew, <0 = left/negative skew).
- **Kurtosis (`data.kurt()`)**: Tail weight and peak sharpness (>0 = heavy-tailed / leptokurtic).

```python
print('--- Distribution Shape ---')
shape = pd.DataFrame({
    'Skewness': data[num_cols].skew().round(4),
    'Kurtosis': data[num_cols].kurt().round(4)
})
shape
```

#### 3.4 Pandas Built-in 5-Number Summary (`describe()`)
Provides count, mean, std, min, 25%, 50% (median), 75%, and max in a single command.

```python
data[num_cols].describe().round(4)
```

#### 3.5 Categorical Descriptive Analytics
Frequency distribution and mode for qualitative variables (`emo_label` and `senti_label`).

```python
print('Emotion Class Counts:')
print(data['emo_label'].value_counts())
print('\nSentiment Counts:')
print(data['senti_label'].value_counts())
```

### Step 4: Graphical Descriptive Analytics

#### 4.1 Histogram (Data Distribution)
Shows the frequency distribution of tweet word counts with mean and median lines.

```python
plt.figure(figsize=(8, 4.5), dpi=150)
data['word_count'].hist(bins=25, color='#2563EB', edgecolor='black', alpha=0.75)
plt.title('Histogram: Distribution of Tweet Word Counts', fontsize=12, fontweight='bold')
plt.xlabel('Word Count', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.axvline(data['word_count'].mean(), color='red', linestyle='--', linewidth=1.5, label=f'Mean ({data["word_count"].mean():.2f})')
plt.axvline(data['word_count'].median(), color='green', linestyle='-', linewidth=1.5, label=f'Median ({data["word_count"].median():.2f})')
plt.legend()
plt.show()
```

#### 4.2 Box Plot (Spread & Outlier Identification)
Visualizes the 5-number summary (Min, Q1, Median, Q3, Max) and highlights statistical outliers in intraday price range.

```python
plt.figure(figsize=(7, 4.5), dpi=150)
data.boxplot(column='intraday_range', patch_artist=True,
             boxprops=dict(facecolor='#93C5FD', color='#1E3A8A'),
             medianprops=dict(color='red', linewidth=2),
             flierprops=dict(marker='o', color='orange', alpha=0.5))
plt.title('Box Plot: Intraday Stock Price Range & Outliers', fontsize=12, fontweight='bold')
plt.ylabel('Intraday Range ((High - Low) / Low)', fontsize=10)
plt.show()
```

#### 4.3 Bar Chart (Categorical Comparison)
Compares post counts across the 12 fine-grained investor emotion categories.

```python
plt.figure(figsize=(9, 4.5), dpi=150)
data['emo_label'].value_counts().plot(kind='bar', color='#10B981', edgecolor='black', width=0.6)
plt.title('Bar Chart: Frequency of 12 Emotion Classes', fontsize=12, fontweight='bold')
plt.xlabel('Emotion Category', fontsize=10)
plt.ylabel('Post Count', fontsize=10)
plt.xticks(rotation=35, ha='right')
plt.show()
```

#### 4.4 Line Chart (Trend Over Time)
Tracks daily average stock price returns across calendar year 2020.

```python
plt.figure(figsize=(10, 4.5), dpi=150)
data['date'] = pd.to_datetime(data['date'])
daily_avg_ret = data.groupby('date')['same_day_return'].mean() * 100
plt.plot(daily_avg_ret.index, daily_avg_ret.values, color='#8B5CF6', linewidth=1.5, label='Daily Average Return (%)')
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
plt.title('Line Chart: Daily Average Stock Return (%) Trend in 2020', fontsize=12, fontweight='bold')
plt.xlabel('Date', fontsize=10)
plt.ylabel('Average Return (%)', fontsize=10)
plt.legend()
plt.show()
```
